## 1. Project Introduction

**CARE (Customer Analysis & Response Engine)** is a Python-based
customer feedback intelligence pipeline designed to identify
critical customer reviews and automate personalized customer-support
outreach.

The system combines **data wrangling, transparent rule-based
filtering, complaint keyword analysis, and Generative AI** to
transform raw customer feedback into actionable customer-support
responses.

The project follows a three-stage analytical workflow:

1. **Data Wrangling**  
   Prepare, validate, and clean the customer review dataset so that
   it is suitable for reliable analysis.

2. **Rule-Based Complaint Detection & Insights**  
   Identify critical customer reviews using transparent business
   rules and analyze recurring complaint keywords and themes.

3. **Generative AI Outreach**  
   Select three of the most critical and detailed negative reviews
   and use a Generative AI API to generate short, personalized, and
   empathetic customer-support apology emails.

The primary design principle of CARE is **explainability**. The
system will use clearly defined Python rules for identifying
critical feedback rather than relying on an opaque Machine Learning
model. Generative AI will be used specifically for drafting
customer-facing responses after the critical cases have been
identified.

This approach creates a traceable workflow from the original
customer review to the final automated response.

## 2. Business Problem

Retail businesses receive a large volume of customer reviews that
contain valuable information about product quality, fit, appearance,
customer expectations, and overall satisfaction.

When this feedback is processed manually, customer-support teams
must spend considerable time reading reviews, identifying the most
urgent negative cases, understanding the customer's complaint, and
preparing individual responses.

This creates three key challenges:

- **Volume:** A large number of reviews makes manual identification
  of critical feedback time-consuming.

- **Prioritization:** Not every negative review requires the same
  level of attention, making it necessary to identify the most
  critical cases using consistent rules.

- **Response Generation:** Once a critical complaint is identified,
  preparing a personalized and empathetic response for every
  customer requires additional manual effort.

CARE addresses these challenges by creating a structured pipeline
that first identifies critical feedback using transparent,
rule-based logic and then uses Generative AI to assist with the
creation of personalized customer-support responses.

### 12.5 CARE Text Preprocessing Rules

The preprocessing function is designed specifically for
rule-based customer complaint detection.

The following transformations will be applied:

1. Convert text to lowercase so that complaint keywords can be
   matched consistently regardless of capitalization.

2. Normalize whitespace so that unnecessary spaces and line breaks
   do not affect matching.

3. Preserve alphabetic words, numbers, and meaningful word
   boundaries.

4. Preserve negation words such as `not`, `never`, `don't`,
   `didn't`, and similar expressions because they can reverse or
   modify the meaning of customer feedback.

5. Do not remove stopwords indiscriminately because common words can
   contribute to complaint expressions.

6. Do not stem or aggressively lemmatize the reviews because the
   original wording may contain useful complaint-specific phrases.

7. Preserve the original customer review separately so that the
   cleaned representation is always traceable back to the source.

The resulting cleaned text will be used for rule-based analysis,
while the original review text will remain available for human
interpretation and Generative AI outreach.

```python
# Import Python's regular-expression library.
# It will be used to normalize whitespace and remove unwanted
# formatting noise without aggressively altering the customer's
# wording.

import re
```

### 12.7 CARE Review Text Cleaning Function

The following function applies the CARE preprocessing rules to an
individual review.

The function returns an empty string when no review text is available.
For available text, it converts the content to lowercase and
normalizes whitespace while preserving the customer's meaningful
words.

The function intentionally avoids aggressive stopword removal,
stemming, and lemmatization because CARE relies on complaint phrases
and negation patterns for rule-based detection.

```python
def clean_review_text(text):
    """
    Clean customer review text for CARE's rule-based complaint analysis.

    The function standardizes formatting while preserving the
    customer's meaningful language and complaint-related context.
    """

    # Return an empty string when Review Text is missing.
    # Missing reviews cannot contribute textual evidence.
    if pd.isna(text):
        return ""

    # Convert the review to lowercase so that keyword matching
    # becomes case-insensitive.
    text = text.lower()

    # Replace line breaks, tabs, and repeated whitespace with
    # a single regular space.
    text = re.sub(r"\s+", " ", text)

    # Remove leading and trailing whitespace after normalization.
    return text.strip()
```

### 12.8 Creating the Clean Review Text Field

The preprocessing function is now applied to every record in the
working dataset.

The cleaned representation will be stored in a new column named
`Clean Review Text`.

The original `Review Text` and `Original Review Text` columns will

This creates a traceable relationship between source feedback and
the representation used for rule-based complaint detection.

```python
# Apply the CARE text-cleaning function to every review.
# The result is stored separately so that the original customer
# feedback remains unchanged.

df["Clean Review Text"] = df["Review Text"].apply(clean_review_text)

print("Clean Review Text column created.")
```

### 14.2 Complaint Keyword Definition

### 14.3 Keyword Counting Function

The following function uses basic Python text processing and
`Counter`, as suggested by the internship assessment.

For every Critical Review, the cleaned review text is tokenized and
matching complaint keywords are counted.

The function returns the most frequently occurring complaint
keywords across the Critical Review population.


```python
def find_common_complaint_keywords(
    review_texts,
    complaint_keywords,
    top_n=15
):
    """
    Count complaint-related keyword occurrences in customer reviews.

    Parameters
    ----------
    review_texts : pandas.Series
        Clean customer review text.

    complaint_keywords : list
        Complaint-related vocabulary to count.

    top_n : int
        Number of most common keywords to return.

    Returns
    -------
    pandas.DataFrame
        Keyword frequency table.
    """

    keyword_set = set(complaint_keywords)
    keyword_counter = Counter()

    for text in review_texts.fillna(""):
        tokens = re.findall(r"\b[a-z]+\b", text.lower())

        matching_keywords = [
            token for token in tokens
            if token in keyword_set
        ]

        keyword_counter.update(matching_keywords)

    return pd.DataFrame(
        keyword_counter.most_common(top_n),
        columns=["Complaint Keyword", "Frequency"]
    )
```

```python
# Calculate the most common complaint keywords in Critical Reviews.

top_complaint_keywords = find_common_complaint_keywords(
    critical_reviews_df["Clean Review Text"],
    complaint_keyword_list
)

top_complaint_keywords
```

### 16.6 Customer Support Agent Prompt

The Gemini model is instructed to act as a Customer Support Agent and generate a short, personalized, and empathetic apology email for each selected Critical Review.


```python
CUSTOMER_SUPPORT_PROMPT = """
You are a Customer Support Agent for an online retail company.

Read the customer's review carefully and write a short,
personalized, and empathetic apology email.

Requirements:
- Acknowledge the customer's specific negative experience.
- Address the specific complaint or complaints mentioned in the review.
- Be polite, professional, and empathetic.
- Keep the response short and clear.
- Do not argue with the customer.
- Do not blame the customer.
- Do not invent facts about the company, product, order, refund,
  replacement, discount, or company policies.
- Do not make unsupported promises.
- Do not mention that an AI model was used.
- Return only the customer-facing apology email.

Customer review:
{review}
"""

print("Customer Support Agent prompt created successfully.")
```

### 16.7 Generate the 3 AI-Generated Response Emails

The three selected Critical Reviews are now passed individually to the Gemini Generative AI API.

Each review is processed separately so that the generated response directly addresses that customer's specific feedback.

Exactly three responses will be generated, matching the three reviews selected in Task 3.


```python
# Generate exactly one customer-facing response for each of the
# three selected Critical Reviews.
#
# This cell is the only part of the notebook that sends customer-review
# text to the Generative AI API.
#
# A short retry mechanism is included only for temporary 503 server
# availability errors. Other API errors are allowed to surface normally.

import time
from google.genai.errors import ServerError

generated_responses = []

for review_number, (_, row) in enumerate(selected_reviews_df.iterrows(), start=1):
    # Insert the customer's original review into the support-agent prompt.
    prompt = CUSTOMER_SUPPORT_PROMPT.format(
        review=row["Original Review Text"]
    )

    for attempt in range(3):
        try:
            response = client.models.generate_content(
                model="gemini-3.1-flash-lite",
                contents=prompt
            )
            break
        except ServerError as error:
            if getattr(error, "status_code", None) != 503 or attempt == 2:
                raise
            wait_seconds = 10 * (2 ** attempt)
            print(f"Temporary Gemini error. Retrying in {wait_seconds} seconds...")
            time.sleep(wait_seconds)

    if not response.text:
        raise ValueError(f"No response returned for review {review_number}.")

    generated_responses.append(response.text.strip())

print("AI responses generated successfully:", len(generated_responses))

if len(generated_responses) != 3:
    raise ValueError("Exactly 3 AI-generated response emails are required.")
```

### 16.8 AI-Generated Customer Support Emails

The next code cell renders the **actual three responses returned by Gemini as Markdown**.

The responses are generated from the three selected customer reviews and are not manually invented or hard-coded into the notebook.

The same three outputs are also saved automatically as `AI_Generated_Responses.md`.

```python
# Display the three AI-generated customer-support emails as Markdown.
#
# The generated responses are displayed in rendered Markdown format and
# are also saved to a companion Markdown file for submission/reference.

from IPython.display import Markdown, display
from pathlib import Path

if len(generated_responses) != 3:
    raise ValueError("Exactly three generated responses are required before display.")

markdown_sections = []

for i, (_, row) in enumerate(selected_reviews_df.iterrows(), start=1):
    response_text = generated_responses[i - 1]
    review_text = str(row["Original Review Text"]).replace("\n", "\n> ")

    markdown_sections.append(
        f"### AI-Generated Response {i}\n\n"
        f"**Clothing ID:** {row['Clothing ID']}  \n"
        f"**Rating:** {row['Rating']}  \n"
        f"**Title:** {row['Title']}\n\n"
        f"**Customer Review:**\n\n"
        f"> {review_text}\n\n"
        f"**Customer Support Email:**\n\n"
        f"{response_text}\n"
    )

display(Markdown("\n---\n".join(markdown_sections)))

# Save the three actual generated emails as a separate Markdown artifact.
responses_md_path = Path("AI_Generated_Responses.md")
responses_md_path.write_text(
    "# AI-Generated Customer Support Emails\n\n"
    + "\n---\n".join(markdown_sections),
    encoding="utf-8"
)

print("Saved Markdown output artifact:", responses_md_path)
```
